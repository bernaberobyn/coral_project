# Function to split the dataset into train, val, and test

```python
import os
import random
import shutil
from pathlib import Path


def split_dataset(data_dir, output_dir, train_count, val_count):
    """Splits a dataset into train, val, and test folders based on absolute counts.

    The remaining images automatically go to the test set.
    """
    data_dir = Path(data_dir)
    output_dir = Path(output_dir)

    # Define target directories
    splits = ["train", "val", "test"]

    # Get all category folders (subdirectories)
    categories = [f for f in data_dir.iterdir() if f.is_dir()]

    for category in categories:
        category_name = category.name
        print(f"Processing category: {category_name}")

        # Get and shuffle all images in the category
        # Add/remove extensions in the glob if you have different formats
        images = [
            f
            for f in category.iterdir()
            if f.suffix.lower() in [".jpg", ".jpeg", ".png", ".bmp"]
        ]
        random.shuffle(images)

        total_images = len(images)
        print(f"  Total images found: {total_images}")

        # Safety check to ensure you aren't requesting more images than exist
        if train_count + val_count > total_images:
            print(
                f"  [Warning] Requested train ({train_count}) + val ({val_count}) "
                f"is greater than total images ({total_images}). Skipping this category."
            )
            continue

        # Split the list based on your counts
        train_images = images[:train_count]
        val_images = images[train_count : train_count + val_count]
        test_images = images[train_count + val_count :]

        print(
            f"  Splits -> Train: {len(train_images)}, Val: {len(val_images)}, Test: {len(test_images)}"
        )

        # Map splits to their respective image lists
        split_mapping = {
            "train": train_images,
            "val": val_images,
            "test": test_images,
        }

        # Create directories and copy files
        for split, split_imgs in split_mapping.items():
            # Create the specific folder (e.g., output/train/category_name)
            target_folder = output_dir / split / category_name
            target_folder.mkdir(parents=True, exist_ok=True)

            # Copy images
            for img in split_imgs:
                shutil.copy(img, target_folder / img.name)

    print("\nDataset splitting complete!")
```

# Split the dataset into train, val, and test

```python
# --- CONFIGURATION ---
# Path to your current dataset containing the 2 category folders
SOURCE_DATASET = "/kaggle/input/datasets/vencerlanz09/healthy-and-bleached-corals-image-classification"

# Where you want the new train/val/test folders to be created
OUTPUT_DATASET = "/kaggle/working/"

# Specify the exact number of images per category
NUM_TRAIN = 352
NUM_VAL = 43

# Run the split
split_dataset(SOURCE_DATASET, OUTPUT_DATASET, NUM_TRAIN, NUM_VAL)
```
